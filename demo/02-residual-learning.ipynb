```python
import numpy as np
import torch
import matplotlib.pyplot as plt
```

```python
class ResidualBlock(torch.nn.Module):
    def __init__(self,channels):
        super().__init__()
        self.conv1 = torch.nn.Conv2d(channels, channels, (3,3,), padding=1)
        self.batchNorm1 = torch.nn.BatchNorm2d(channels)
        self.relu = torch.nn.ReLU(inplace=True)
        self.conv2 = torch.nn.Conv2d(channels,channels, (3,3,), padding=1)
        self.batchNorm2 = torch.nn.BatchNorm2d(channels)
    
    def forward(self, X):
        out = self.conv1(X)
        out = self.batchNorm1(out)
        self.relu(out)
        
        out = self.conv2(out)
        out = self.batchNorm2(out)
        out += X
        self.relu(out)
        
        return out        
```

```python
resid = ResidualBlock(channels=64)

x = torch.rand((1,64,56,56,))
y = resid.forward(x)

print(f"Input shape: {x.shape}")
print(f"Output shape: {y.shape}")
```
