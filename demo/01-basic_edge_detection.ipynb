```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
kernel = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
img = Image.open('D:\\Adobe\\random\\aankhein.png')
img_np = np.asarray(img)[:,:,:3] #discarding alpha(opacity) in RGBA
img_grey = np.mean(img_np,axis=2)
img_grey.shape
```

```python
img_pad = np.pad(img_grey,20)
output = []
kernel_flat = kernel.flatten()
for i in range(img_pad.shape[0]-2):
    row = []
    for j in range(img_pad.shape[1]-2):
        prod = np.dot(kernel_flat,img_pad[i:3+i,j:3+j].flatten())
        row.append(prod)
    output.append(row)
out_np = np.array(output)
out_np.shape
```

```python
fig, axes = plt.subplots(nrows=1,ncols=2,figsize=(15,300))
plt.subplot(1,2,1)
plt.imshow(out_np,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img_grey,cmap='gray')
```
