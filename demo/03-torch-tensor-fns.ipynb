```python
import torch
```

```python
# X is (no of images[batch size],channels[rgb],height,width)
X = torch.rand((1,3,224,224,))
X_unfold = X.unfold(2,16,16).unfold(3,16,16)
X_unfold.shape # returns (batch size,channels,no of patches(x),y,
                        # height of each patch, width)
```

```python
X_cont = torch.permute(X_unfold, (0,2,3,1,4,5,)).contiguous()
X_flat = torch.flatten(X_cont, start_dim=1, end_dim=2)
X_flat_final = torch.flatten(X_flat, start_dim=2, end_dim=-1)
X_flat_final.shape
```

```python
model = torch.nn.Linear(in_features=768,out_features=768)
y = model(X_flat_final)
y.shape
```
